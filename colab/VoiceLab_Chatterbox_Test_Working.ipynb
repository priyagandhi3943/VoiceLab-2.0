```python
import torch
from chatterbox.tts import ChatterboxTTS

device = "cuda" if torch.cuda.is_available() else "cpu"

model = ChatterboxTTS.from_pretrained(device=device)

print("Chatterbox model loaded successfully!")
```

```python
from google.colab import files

uploaded = files.upload()

REFERENCE_AUDIO = list(uploaded.keys())[0]

print("Voice recording uploaded:", REFERENCE_AUDIO)
```

```python
import torchaudio
from IPython.display import Audio, display

TEXT = "Hello , welcome I am Megh.This is a demonstration of voice cloning using artificial intelligence. I love football. Priya is the best. "

wav = model.generate(
    TEXT,
    audio_prompt_path=REFERENCE_AUDIO
)

torchaudio.save(
    "chatterbox_generated.wav",
    wav.cpu(),
    model.sr
)

print("Voice generated successfully!")
display(Audio("chatterbox_generated.wav"))
```
